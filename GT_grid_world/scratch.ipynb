```python
import numpy as np

np.random.seed(0)

Ra = []

# Find the maximum number of locations of all tasks
l_max = 20
# Get the number of tasks
num_tasks = 2

num_robots = 8

arrays = []

for robot in range(num_robots):
    arrays.append(np.random.random((l_max, num_tasks))*10)

cost_tensor = np.dstack(arrays)

print(cost_tensor.shape)

print(np.min(cost_tensor))
print(np.unravel_index(np.argmin(cost_tensor), cost_tensor.shape))

min_value_indexes = []

for _ in range(num_robots):
    min_value_indexes.append(np.unravel_index(np.argmin(cost_tensor), cost_tensor.shape))
    robot = min_value_indexes[-1][2]
    task = min_value_indexes[-1][1]
    
    cost_tensor = np.delete(cost_tensor, robot, 2)
    cost_tensor = np.delete(cost_tensor, task, 1)
    Ra.append((task, robot))
    
    if 0 in cost_tensor.shape:
        break
    
    
print(cost_tensor.shape)
# print(np.min(cost_tensor))

print(Ra)
```
