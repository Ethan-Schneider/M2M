```python
inbound_to_outboud = 9
inbound_probability = inbound_to_outboud/(inbound_to_outboud + 1)
outbound_probability = 1 - inbound_probability
nums = np.random.choice([0, 1], size=10000, p=[outbound_probability, inbound_probability])
print(np.sum(nums))
```
